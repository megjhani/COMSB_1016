## Height ##

### Question: What is the height of the [40, 65] bin?
### In other words, what % of the top movies are between 40-65 years old?

```python
binned_data = top_movies.bin('Age', bins = my_bins)
binned_data
```

```python
total = sum(binned_data.column('Age count'))
total
```

```python
# Add a column containing what percent of movies are in each bin
binned_data = binned_data.with_column(
    'Percent', 100*binned_data.column('Age count')/total)
binned_data
```

```python
top_movies.hist('Age', bins=my_bins, unit='Year')
```

```python
# Step 1: Calculate % of movies in the [40, 65) bin
percent = binned_data.where('bin', 40).column('Percent').item(0)
percent
```

```python
# Step 2: Calculate the width of the 40-65 bin
width = 65 - 40
```

```python
# Step 3: Area of rectangle = height * width
#         --> height = percent / width
height = percent / width
height
```

this height represents the percent per year (the % of top movies came out within this age range)

## Choosing a chart

- **Bar chart:** compare categories.
- **Histogram:** describe the distribution of one numerical variable.
- **Line plot:** show change across an ordered variable.
- **Scatter plot:** examine the relationship between two numerical variables.
